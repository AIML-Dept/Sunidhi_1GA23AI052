```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

def simon(s):
    n = len(s)

    if set(s) == {"0"}:
        print("All-zero secret:", s)
        return

    p = s.index("1")

    qc = QuantumCircuit(2 * n, n)

    qc.h(range(n))

    k = 0

    for j in range(n):
        if j != p:
            qc.cx(j, n + k)

            if s[j] == "1":
                qc.cx(p, n + k)

            k += 1

    qc.h(range(n))
    qc.measure(range(n), range(n))

    sim = AerSimulator()

    counts = sim.run(
        transpile(qc, sim),
        shots=2048
    ).result().get_counts()

    print("Secret:", s)
    print("Measurements:", counts)


simon("1011")
```
