```python
from qiskit import QuantumCircuit

s = "110"
n = 3

oracle = QuantumCircuit(6)

oracle.cx(0, 3)
oracle.cx(1, 3)
oracle.cx(2, 4)

def f(x):
    return (int(x[0]) ^ int(x[1]), int(x[2]), 0)

print("x    f(x)    f(x⊕110)")

for i in range(8):
    x = format(i, "03b")
    xs = "".join(str(int(a) ^ int(b)) for a, b in zip(x, s))
    print(x, "->", f(x), "->", f(xs))
```
