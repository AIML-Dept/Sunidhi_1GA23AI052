```python
import matplotlib.pyplot as plt

n_values = list(range(2, 11))

classical_queries = [2**(n - 1) + 1 for n in n_values]

quantum_queries = [1 for n in n_values]

for n, classical, quantum in zip(
    n_values, classical_queries, quantum_queries
):
    print(
        f"n = {n}: "
        f"Classical = {classical}, "
        f"Quantum = {quantum}"
    )

plt.plot(
    n_values,
    classical_queries,
    marker='o',
    label="Classical Worst Case"
)

plt.plot(
    n_values,
    quantum_queries,
    marker='o',
    label="Quantum"
)

plt.xlabel("Number of Input Bits (n)")
plt.ylabel("Number of Queries")
plt.title("Classical vs Quantum Query Count")
plt.legend()
plt.grid(True)

plt.show()
```
