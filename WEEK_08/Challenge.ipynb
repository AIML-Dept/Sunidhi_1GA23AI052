```python
import random
from itertools import product


def generate_balanced_function(n):
    """
    Generate a random balanced Boolean function.
    """

    inputs = list(product([0, 1], repeat=n))

    total = len(inputs)

    half = total // 2

    outputs = [0] * half + [1] * half

    random.shuffle(outputs)

    function = dict(zip(inputs, outputs))

    return function


def classify_function(function, n):
    """
    Determine whether the function is balanced
    or constant.
    """

    outputs = list(function.values())

    if all(value == 0 for value in outputs):
        return "constant"

    if all(value == 1 for value in outputs):
        return "constant"

    return "balanced"

for n in range(2, 6):

    function = generate_balanced_function(n)

    result = classify_function(function, n)

    print(f"\nn = {n}")
    print("Generated function:")

    for input_value, output_value in function.items():
        print(input_value, "->", output_value)

    print("Classification:", result)

    if result == "balanced":
        print("Correct classification!")
    else:
        print("Incorrect classification!")
```
