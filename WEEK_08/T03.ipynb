```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

def deutsch_jozsa(n, balanced=True):
    qc = QuantumCircuit(n + 1, n)

    qc.x(n)
    qc.h(n)

    qc.h(range(n))

    if balanced:
        for i in range(n):
            qc.cx(i, n)

    qc.h(range(n))

    qc.measure(range(n), range(n))

    return qc

simulator = AerSimulator()

for n in range(2, 6):
    print("\nn =", n)

    qc = deutsch_jozsa(n, False)
    result = simulator.run(
        transpile(qc, simulator),
        shots=100
    ).result()
    print("Constant:", result.get_counts())

    qc = deutsch_jozsa(n, True)
    result = simulator.run(
        transpile(qc, simulator),
        shots=100
    ).result()
    print("Balanced:", result.get_counts())
```
