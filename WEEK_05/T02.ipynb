```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

s = "110"
n = 3

qc = QuantumCircuit(6, 3)

qc.h(range(n))

qc.cx(0, 3)
qc.cx(1, 3)
qc.cx(2, 4)

qc.h(range(n))
qc.measure(range(n), range(n))

result = AerSimulator().run(
    transpile(qc, AerSimulator()),
    shots=1024
).result()

counts = result.get_counts()

print(counts)

print("\nEquations y·s = 0:")
for y in list(counts)[:2]:
    print(y, "·", s, "= 0 (mod 2)")
```
