```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import random
from scipy.stats import chisquare

shots = 10000

# Quantum random generator
qc = QuantumCircuit(1, 1)

qc.h(0)
qc.measure(0, 0)

simulator = AerSimulator()
result = simulator.run(qc, shots=shots).result()
quantum_counts = result.get_counts()

quantum_0 = quantum_counts.get('0', 0)
quantum_1 = quantum_counts.get('1', 0)

# Classical biased random generator
classical_bits = random.choices(
    [0, 1],
    weights=[0.6, 0.4],
    k=shots
)

classical_0 = classical_bits.count(0)
classical_1 = classical_bits.count(1)

# Expected counts for a fair distribution
expected = [shots / 2, shots / 2]

# Chi-square tests
quantum_stat, quantum_p = chisquare(
    [quantum_0, quantum_1],
    expected
)

classical_stat, classical_p = chisquare(
    [classical_0, classical_1],
    expected
)

print("QUANTUM GENERATOR")
print("Counts:", quantum_counts)
print("Chi-square statistic:", quantum_stat)
print("p-value:", quantum_p)

print("\nCLASSICAL BIASED GENERATOR")
print("0 count:", classical_0)
print("1 count:", classical_1)
print("Chi-square statistic:", classical_stat)
print("p-value:", classical_p)

alpha = 0.05

print("\nINTERPRETATION")

if quantum_p < alpha:
    print("Quantum generator appears significantly biased.")
else:
    print("Quantum generator is consistent with a fair random source.")

if classical_p < alpha:
    print("Classical generator is significantly biased.")
else:
    print("Classical generator is consistent with a fair random source.")
```
