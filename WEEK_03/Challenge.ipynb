```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc1 = QuantumCircuit(1)
qc1.h(0)
qc1.z(0)
qc1.h(0)

qc2 = QuantumCircuit(1)
qc2.x(0)

print("Circuit 1 (HZH):")
print(qc1.draw())

print("\nCircuit 2 (X):")
print(qc2.draw())

state1 = Statevector.from_instruction(qc1)
state2 = Statevector.from_instruction(qc2)

print("\nStatevector after HZH:")
print(state1)

print("\nStatevector after X:")
print(state2)

print("\nAnalytical Proof:")
print("HZH = X")
print("Applying HZH on |0⟩ gives |1⟩, which is identical to applying X on |0⟩.")

print("\nVerification:")
if state1.equiv(state2):
    print("Success! HZH is equivalent to the X gate.")
else:
    print("Verification Failed!")
```
