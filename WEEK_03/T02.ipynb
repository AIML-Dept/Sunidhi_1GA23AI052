```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

qc.h(0)
qc.z(0)

print("Quantum Circuit:")
print(qc.draw())

state = Statevector.from_instruction(qc)

print("\nStatevector of the |-> state:")
print(state)

expected = Statevector([1/2**0.5, -1/2**0.5])

print("\nExpected Statevector:")
print(expected)

print("\nVerification:")

if state.equiv(expected):
    print("Success! The circuit correctly prepares the |-> state.")
else:
    print("Verification Failed!")
```
