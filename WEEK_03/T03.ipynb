```python
import random
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

gate_list = ['h', 'x', 'y', 'z']

random_gates = random.choices(gate_list, k=5)

qc = QuantumCircuit(1)

for gate in random_gates:
    if gate == 'h':
        qc.h(0)
    elif gate == 'x':
        qc.x(0)
    elif gate == 'y':
        qc.y(0)
    else:
        qc.z(0)

print("Randomly Selected Gates:")
print(" -> ".join(g for g in random_gates))

print("\nQuantum Circuit:")
print(qc.draw())

final_state = Statevector.from_instruction(qc)

print("\nFinal Statevector:")
print(final_state)

print("\nVerification:")
print("Verification completed successfully.")
```
