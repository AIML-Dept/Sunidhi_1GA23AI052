```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

qc.h(0)

print("Quantum Coin Flip Circuit:")
print(qc.draw())

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

probabilities = state.probabilities()

print("\nProbability of Heads (|0>):", probabilities[0])
print("Probability of Tails (|1>):", probabilities[1])

if probabilities[0] > probabilities[1]:
    print("\nWinner: Heads")
elif probabilities[1] > probabilities[0]:
    print("\nWinner: Tails")
else:
    print("\nResult: Fair Coin (Equal chance of Heads and Tails)")

print("\nVerification: Successful")
```
