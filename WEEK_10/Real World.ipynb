```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
import random

simulator = AerSimulator()

messages = [random.choice(["00", "01", "10", "11"]) for _ in range(100)]

classical_bits = 0
superdense_qubits = 0
correct = 0

for message in messages:
    qc = QuantumCircuit(2, 2)

    qc.h(0)
    qc.cx(0, 1)

    if message == "01":
        qc.x(0)
    elif message == "10":
        qc.z(0)
    elif message == "11":
        qc.x(0)
        qc.z(0)

    qc.cx(0, 1)
    qc.h(0)

    qc.measure([0, 1], [0, 1])

    compiled = transpile(qc, simulator)
    result = simulator.run(compiled, shots=1).result()
    decoded = list(result.get_counts().keys())[0]

    classical_bits += 2
    superdense_qubits += 1

    if decoded == message:
        correct += 1

print("SUPERDENSE CODING COMMUNICATION")
print("--------------------------------")
print("Total messages:", len(messages))
print("Classical bits required:", classical_bits)
print("Qubits transmitted:", superdense_qubits)
print("Correctly decoded messages:", correct)
print("Decoding accuracy:", (correct / len(messages)) * 100, "%")
print()
print("CLASSICAL TRANSMISSION")
print("----------------------")
print("Bits transmitted:", classical_bits)
print()
print("SUPERDENSE TRANSMISSION")
print("-----------------------")
print("Qubits transmitted:", superdense_qubits)
print()
print("Transmission reduction:",
      classical_bits - superdense_qubits, "units")
print("Transmission efficiency:",
      ((classical_bits - superdense_qubits) / classical_bits) * 100, "%")
```
